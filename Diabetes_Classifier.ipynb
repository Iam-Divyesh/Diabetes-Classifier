```python
#importing necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('./pima_data/diabetes.csv')

df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df[(df['BloodPressure'] >= 0) & (df['BloodPressure'] <=40)]
```

### **It is Not possible someone having 0 Blood Pressure**
so that i am removing all that related data

```python
df[df['BloodPressure'] == 0].shape
```

```python
df.shape
```

```python
df1 = df[df['BloodPressure'] != 0]
```

```python
df1.shape
```

```python
df1.describe()
```

```python
df2 = df1[df1['Glucose'] != 0]
df3 = df2[df2['BMI'] != 0]
```

### **Modle Training**

```python
# Spliting Data for training and testing
from sklearn.model_selection import train_test_split
X = df3.drop('Outcome', axis=1)
y = df3['Outcome']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression

logistic = LogisticRegression()
```

```python
logistic.fit(X_train , y_train)
```

```python
y_pred = logistic.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report , confusion_matrix
```

```python
score = accuracy_score(y_test , y_pred)
print(score)
print(confusion_matrix(y_test , y_pred))
print(classification_report(y_test , y_pred))
```

### **Hyperparameter Tuning**
GridSearchCv and StratifiedSearchCv

```python
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import StratifiedKFold
cv = StratifiedKFold()

penalty = ['l1', 'l2', 'elasticnet']

c_values = [100,10,1.0,0.1,0.01]

solver = ['lbfgs', 'liblinear', 'newton-cg', 'newton-cholesky', 'sag', 'saga']

params = dict(penalty = penalty , C = c_values , solver = solver)

grid = GridSearchCV(estimator=logistic , param_grid=params ,scoring='accuracy' ,cv = cv , n_jobs=-1 )
```

```python
grid.fit(X_train , y_train)
```

```python
grid.best_params_
```

```python
y_pred = grid.predict(X_test)
```

```python
score = accuracy_score(y_test , y_pred)
print(score)
print(confusion_matrix(y_test , y_pred))
print(classification_report(y_test , y_pred))
```

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
model = LogisticRegression()

randomcv = RandomizedSearchCV(
    estimator=model,
    param_distributions=params,
    cv = 5,
    scoring= 'accuracy'
)
```

```python
randomcv.fit(X_train , y_train)
```

```python
y_pred = randomcv.predict(X_test)
```

```python
score = accuracy_score(y_test , y_pred)
print(score)
print(confusion_matrix(y_test , y_pred))
print(classification_report(y_test , y_pred))
```
